```python
from dbfread import DBF
```

```python
dataset_path = "../dataset/CAMS AMCs NAVs 09052025/11052025150310_185847860R1.dbf"

# Read the DBF file
dbf = DBF(dataset_path)

# Display the column names and the number of records in the DBF file
print("Columns in the DBF file:", dbf.field_names)
print("Number of records in the DBF file:", len(dbf))
```

```python
# Display the first 5 records from the DBF file
from itertools import islice
sample = iter(islice(dbf, 5))

for record in sample:
    print(record)
```

```python
# Convert the DBF file to a pandas DataFrame
# Read the first 10,000 records for initial analysis
import pandas as pd

df_sample = pd.DataFrame(islice(iter(dbf), 10_000))  
print("Sample DataFrame shape:", df_sample.shape)
print("Sample DataFrame info:")
df_sample.info()
```

```python
for col in ['PRODCODE', 'SCHEME_TYP', 'ISIN_NO', 'SWING_NAV']:
    print(f"Unique values in column '{col}':", df_sample[col].nunique())
```

```python
print("Date Range: ", df_sample['NAV_DATE'].min(), "to", df_sample['NAV_DATE'].max())
```

```python
print("Missing values in each column:")
print(df_sample.isnull().sum())
```

```python
summary_stats = df_sample.describe(include='number')
print("Summary statistics for the sample DataFrame:")
print(summary_stats)
```

```python
unique_values = {}
for col in ['PRODCODE', 'NAME', 'SCHEME_TYP', 'ISIN_NO', 'SWING_NAV']:
    unique_values[col] = set()
    for record in islice(dbf, 100_000):
        unique_values[col].add(record[col])

print("unique valeyes in 100,000 records:")
print("Number of unique products:", len(unique_values['PRODCODE']))
print("Number of unique product names:", len(unique_values['NAME']))
print("Number of unique scheme types:", len(unique_values['SCHEME_TYP']))
print("Number of unique ISIN numbers:", len(unique_values['ISIN_NO']))
print("Number of unique swing NAVs:", len(unique_values['SWING_NAV']))
```

```python
# Analyze the records for a specific product
PRODUCT = df_sample['PRODCODE'][0]

product_records = []
for record in islice(dbf, 100_000):
    if record['PRODCODE'] == PRODUCT:
        product_records.append(record)
    else:
        break

print(f"Number of records for product {PRODUCT} in the first 100,000 records:", len(product_records))
```

```python
B02_DATA = pd.DataFrame(product_records)
print(f"Sample records for product {PRODUCT}:")
B02_DATA.info()
B02_DATA.describe(include='number')
```

```python
for col in ["PRODCODE", "NAME", "SCHEME_TYP", "ISIN_NO"]:
    print(f"{col}: {B02_DATA[col].nunique()}")
```

```python
print("Date range:")
print(B02_DATA["NAV_DATE"].min(), "→", B02_DATA["NAV_DATE"].max())
```

```python
dates = B02_DATA["NAV_DATE"]

print("Dates sorted:", dates.is_monotonic_increasing)
print("Dates sorted descending:", dates.is_monotonic_decreasing)
```
