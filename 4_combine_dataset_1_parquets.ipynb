```python
import pyarrow as pa
import pyarrow.parquet as pq
from pathlib import Path
```

```python
products_dir = Path("../dataset/CAMS AMCs NAVs 09052025/processed/products")
output_file = Path("../dataset/CAMS AMCs NAVs 09052025/processed/all_products.parquet")
```

```python
files = list(products_dir.glob("*.parquet"))

# Fixed schema for the combined file
schema = pa.schema([
    ("PRODCODE", pa.string()),
    ("NAV_DATE", pa.date32()),
    ("NAV_VALUE", pa.float64()),
    ("DIV_PERUNI", pa.float64()),
    ("CORPDIVRAT", pa.float64()),
    ("SWING_NAV", pa.float64())
])
```

```python
writer = pq.ParquetWriter(output_file, schema)

for file in files:
    table = pq.read_table(file)

    table = table.append_column(
        "PRODCODE",
        pa.array([file.stem] * table.num_rows)
    )

    table = table.select(schema.names).cast(schema)

    writer.write_table(table)
    
writer.close()
```

```python
print("Combined file created:", output_file)
```
